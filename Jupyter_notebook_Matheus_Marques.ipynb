## Etapas da Exploração de Dados

Agora que os dados estão carregados, vamos responder às perguntas da atividade usando comandos Python.

### Pergunta 1 — Quantas colunas os dados contêm?

```python
print(f"Número de colunas: {df.shape[1]}")
print("\nNomes das colunas:")
print(df.columns.tolist())
```

### Pergunta 2 — Quantas linhas (amostras)?

```python
print(f"Número de linhas (amostras): {df.shape[0]}")
```

### Pergunta 3 — Que tipos de características existem? Quais são categóricas e quais são numéricas?

```python
print("Tipos de dados de cada coluna:")
print(df.dtypes)

print("\n---")
categorical_cols = df.select_dtypes(include=['object', 'str', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

print(f"\nColunas categóricas/texto ({len(categorical_cols)}): {categorical_cols}")
print(f"\nColunas numéricas ({len(numerical_cols)}): {numerical_cols}")
```

### Pergunta 4 — Qual é a aparência dos dados segundo essas características?

Para características numéricas, examinamos o intervalo de valores com `.describe()`. Para características categóricas, examinamos a frequência de cada classe.

```python
# Resumo estatístico das colunas numéricas
df.describe()
```

```python
# Frequência de valores em colunas categóricas (se existirem)
if categorical_cols:
    for col in categorical_cols:
        print(f"\nColuna '{col}':")
        print(df[col].value_counts())
else:
    print("Não há colunas do tipo object/category.")
    print("\nVerificando colunas numéricas com valores discretos (possíveis categóricas codificadas):")
    for col in df.columns:
        n_unique = df[col].nunique()
        if n_unique <= 10:
            print(f"  '{col}' — {n_unique} valores únicos: {sorted(df[col].dropna().unique().tolist())}")
```

### Pergunta 5 — Existem dados faltando?

```python
missing = df.isnull().sum()
total_missing = missing.sum()

if total_missing > 0:
    print("Colunas com dados faltando:")
    print(missing[missing > 0])
    print(f"\nTotal de valores faltando: {total_missing}")
else:
    print("Nenhum dado faltando no dataset.")
```
