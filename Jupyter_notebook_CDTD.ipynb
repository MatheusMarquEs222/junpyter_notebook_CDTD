### Pergunta 3 — Que tipos de características existem? Quais são categóricas e quais são numéricas?

```python
print("Tipos de dados de cada coluna:")
print(df.dtypes)

print("\n---")
categorical_cols = df.select_dtypes(include=['object', 'str', 'category']).columns.tolist()
numerical_cols = df.select_dtypes(include=['number']).columns.tolist()

print(f"\nColunas categóricas/texto ({len(categorical_cols)}): {categorical_cols}")
print(f"\nColunas numéricas ({len(numerical_cols)}): {numerical_cols}")
```

### Pergunta 4 — Qual é a aparência dos dados segundo essas características?

Para características numéricas, examinamos o intervalo de valores com `.describe()`. Para características categóricas, examinamos a frequência de cada classe.

```python
# Resumo estatístico das colunas numéricas
df.describe()
```

```python
# Frequência de valores em colunas categóricas (se existirem)
if categorical_cols:
    for col in categorical_cols:
        print(f"\nColuna '{col}':")
        print(df[col].value_counts())
else:
    print("Não há colunas do tipo object/category.")
    print("\nVerificando colunas numéricas com valores discretos (possíveis categóricas codificadas):")
    for col in df.columns:
        n_unique = df[col].nunique()
        if n_unique <= 10:
            print(f"  '{col}' — {n_unique} valores únicos: {sorted(df[col].dropna().unique().tolist())}")
```

### Pergunta 5 — Existem dados faltando?

```python
missing = df.isnull().sum()
total_missing = missing.sum()

if total_missing > 0:
    print("Colunas com dados faltando:")
    print(missing[missing > 0])
    print(f"\nTotal de valores faltando: {total_missing}")
else:
    print("Nenhum dado faltando no dataset.")
```

# Parte 2 — Explorando as Características Financeiras do Dataset

Nesta segunda parte, aprofundamos a análise nas colunas financeiras do dataset: valores de fatura (`BILL_AMT`) e valores de pagamento (`PAY_AMT`). Usaremos `.describe()`, histogramas e transformações logarítmicas para entender a distribuição desses dados.

---
## Exercício 1 — Listas com nomes das características financeiras

Criamos duas listas separadas: uma para as colunas de **valor da fatura** (bill amounts) e outra para as colunas de **valor do pagamento** (payment amounts).

```python
bill_feats = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3',
              'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']

pay_feats  = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3',
              'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']

print('Características de valor da fatura:', bill_feats)
print('Características de valor do pagamento:', pay_feats)
```

---
## Exercício 2 — Síntese estatística das características de valor da fatura

O `.describe()` retorna métricas como média, desvio padrão, mínimo, máximo e quartis. Observando os resultados, nota-se que **os valores mínimos são negativos**, o que pode parecer incomum, mas faz sentido no contexto de cartão de crédito: um saldo negativo indica que o cliente realizou um pagamento superior ao valor devido, gerando um crédito na conta. Os valores máximos são bastante altos, indicando a presença de clientes com faturas muito elevadas — possíveis outliers que poderão exigir tratamento posterior.

```python
df[bill_feats].describe()
```

---
## Exercício 3 — Histogramas das características de valor da fatura (grade 2×3)

Visualizamos a distribuição de cada coluna de fatura em uma grade de 2 linhas × 3 colunas, com 20 bins cada. A assimetria à direita é clara em todas as colunas: a maioria dos saldos de fatura se concentra em valores mais baixos, com uma cauda longa de valores altos.

```python
import matplotlib.pyplot as plt

df[bill_feats].hist(bins=20, layout=(2, 3), figsize=(14, 8))
plt.suptitle('Distribuição das Características de Valor da Fatura', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```

---
## Exercício 4 — Síntese estatística das características de valor do pagamento

Diferentemente das faturas, os valores de pagamento são **sempre não negativos** (não é possível pagar um valor negativo). A média e a mediana (50%) revelam que a maior parte dos pagamentos é relativamente baixa, mas o valor máximo é muito alto — indicando forte assimetria positiva (cauda longa à direita). A grande proporção de zeros reforça que muitos titulares não efetuaram pagamento em determinados meses.

```python
df[pay_feats].describe()
```

---
## Exercício 5 — Histogramas das características de valor do pagamento (com `xrot`)

A distribuição é **fortemente assimétrica à direita**: a grande maioria dos valores está próxima de zero, com uma barra muito alta na extremidade esquerda e uma cauda longa de valores altos. O argumento `xrot=45` rotaciona os rótulos do eixo x, evitando sobreposição. Essa distribuição assimétrica motiva o uso de transformação logarítmica no Exercício 7.

```python
df[pay_feats].hist(bins=20, layout=(2, 3), figsize=(14, 8), xrot=45)
plt.suptitle('Distribuição das Características de Valor do Pagamento', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```

---
## Exercício 6 — Máscara booleana: quantos pagamentos são exatamente zero?

Usamos uma **máscara booleana** para identificar quais registros possuem valor de pagamento igual a zero. O resultado é consistente com o histograma anterior: a barra inicial era muito alta, confirmando que uma parcela significativa dos clientes não efetuou pagamento em cada mês do período analisado.

```python
zero_mask = df[pay_feats] == 0

print('Quantidade de pagamentos iguais a zero por coluna:')
print(zero_mask.sum())
print(f'\nLinhas com ao menos um pagamento zero: {zero_mask.any(axis=1).sum()}'
      f' ({zero_mask.any(axis=1).mean()*100:.1f}% do total)')
```

---
## Exercício 7 — Histogramas com transformação log₁₀ (pagamentos ≠ 0)

Aplicamos `np.log10()` via `.apply()` apenas nos pagamentos **diferentes de zero**, pois log₁₀(0) é indefinido. A transformação logarítmica comprime a escala, revelando a distribuição que antes ficava oculta pela assimetria. Os histogramas resultantes mostram uma distribuição **aproximadamente normal** na escala log, com a maioria dos pagamentos concentrada entre 10² e 10⁴ (100 a 10.000). Isso indica que a distribuição original é **log-normal** — padrão comum em dados financeiros.

```python
import numpy as np

# Seleciona apenas linhas onde todos os pagamentos são diferentes de zero
nonzero_mask = (df[pay_feats] != 0).all(axis=1)
pay_nonzero  = df.loc[nonzero_mask, pay_feats]

print(f'Registros com todos os pagamentos != 0: {nonzero_mask.sum()}'
      f' ({nonzero_mask.mean()*100:.1f}% do total)')

# Aplica log10 elemento a elemento e plota
pay_nonzero.apply(np.log10).hist(
    bins=20,
    layout=(2, 3),
    figsize=(14, 8),
    color='dimgray',
    edgecolor='white'
)
plt.suptitle('Transformações Logarítmicas dos Pagamentos (≠ 0)', y=1.02, fontsize=14)
plt.tight_layout()
plt.show()
```
